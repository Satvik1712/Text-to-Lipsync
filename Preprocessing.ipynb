```python
import os

def extract_audio(video_path, audio_path):
    os.makedirs(os.path.dirname(audio_path), exist_ok=True)
    command = f'ffmpeg -i "{video_path}" -q:a 0 -map a "{audio_path}" -y'
    os.system(command)
    
    if os.path.exists(audio_path):
        print(f"Audio saved: {audio_path}")
    else:
        print(f"Failed to save audio: {audio_path}")

def extract_frames(video_file, frames_output_dir, frame_rate=25):
    os.makedirs(frames_output_dir, exist_ok=True)
    command = f'ffmpeg -i "{video_file}" -vf fps={frame_rate} "{frames_output_dir}/frame_%03d.jpg"'
    os.system(command)

def process_videos(video_folder, output_root):
    # Ensure the output root folder exists
    os.makedirs(output_root, exist_ok=True)
    
    # Loop through the video files in the folder
    for video_file in os.listdir(video_folder):
        if video_file.endswith(".mpg"):  # Change this if your video files have a different extension
            video_path = os.path.join(video_folder, video_file)
            
            # Generate video ID from filename
            video_id = os.path.splitext(video_file)[0].zfill(5)  # Zero-fill to 5 digits
            
            # Define paths for audio and frames
            audio_path = os.path.join(output_root, video_id, "audio.wav")
            frames_output_dir = os.path.join(output_root, video_id)
            
            # Extract audio
            extract_audio(video_path, audio_path)
            print(f"Processed audio: {video_file} -> {audio_path}")
            
            # Extract frames
            extract_frames(video_path, frames_output_dir)
            print(f"Extracted frames for: {video_file} -> {frames_output_dir}")

# Paths
video_folder = "D:/STLS/s1"
output_root = "D:/STLS/preprocessed"

# Process videos to extract audio and frames
process_videos(video_folder, output_root)

print("Processing complete!")
```

```python
import os
import random

# Path to the video folder
video_folder = "D:/STLS/preprocessed"

# Get list of folders in the video folder
folder_names = [name for name in os.listdir(video_folder) if os.path.isdir(os.path.join(video_folder, name))]

# Shuffle the folder names
random.shuffle(folder_names)

# Split the folder names into train, val, and test sets
train_folders = folder_names[:800]   # First 800 for training
val_folders = folder_names[800:900]  # Next 100 for validation
test_folders = folder_names[900:]     # Last 100 for testing

# Write folder names to text files
with open("train.txt", "w") as train_file:
    for folder in train_folders:
        train_file.write(f"{folder}\n")

with open("val.txt", "w") as val_file:
    for folder in val_folders:
        val_file.write(f"{folder}\n")

with open("test.txt", "w") as test_file:
    for folder in test_folders:
        test_file.write(f"{folder}\n")

print("Folder names have been written to train.txt, val.txt, and test.txt.")
```
